```python
#MONITORING DAN EVALUASI PERFORMA TIM OLAHRAGA

#1. ANALISIS KONDISI DENGAN PERCABANGAN

print("====Evaluasi Kondisi Fisik====")

    #Menggunakan integer dan float untuk menginput angka
hristirahat = int(input("Detak Jantung Istirahat (bpm) :"))
sistolik = int(input("Tekanan Darah Sistolik        :"))
diastolik = int(input("Tekanan Darah Diastolik       :"))
vo2max = float(input("VO2Max                        :"))
imt = float(input("IMT                           :"))

    #Menggunakan percabangan if, elif, dan else untuk menentukan kondisi heart rate
if hristirahat < 60:
    sht = "Bradikardia"
elif hristirahat <= 100:
    sht = "Normal"
else:
    sht = "Takikardia"

    #Menggunakan percabangan if, elif, dan else untuk menentukan kondisi tekanan darah menggunakan logika AND & OR
    #AND menjadi true ketika 2 variabel terpenuhi, sedangkan OR hanya membutuh salah satu saja
if sistolik < 120 and diastolik < 80:
    std = "Normal"
elif sistolik < 139 and diastolik < 89:
    std = "Prehipertensi"
elif sistolik < 159 and diastolik < 99:
    std = "Hipertensi Level 1"
else:
    std = "Hipertensi Level 2"

    #Menggunakan percabangan if, elif, dan else untuk menentukan kondisi VO2Max
if vo2max < 33.0:
    svo = "Sangat Buruk"
elif vo2max <= 36.4:
    svo = "Buruk"
elif vo2max <= 42.4:
    svo = "Cukup"
elif vo2max <= 46.4:
    svo = "Baik"
elif vo2max <= 52.4:
    svo = "Sangat Baik"
else:
    svo = "Superior"

    #Menggunakan percabangan if, elif, dan else untuk menentukan kondisi IMT atlet

if imt < 18.5:
    simt = "Underweight"
elif imt < 24.9:
    simt = "Normal"
elif imt < 29.9:
    simt = "Overweight"
else:
    simt = "Obesitas"

print("\n")
print("-"*50)

#2. PENGOLAHAN DATA BERULANG DENGAN MENGGUNAKAN LOOP (WHILE & FOR)

print("====Input Progres Latihan Lari Mingguan====")
print("-"*50)

    #Buat list kosong untuk menyimpan riwayat lari
riwayatlatihan = []

    #Menggunakan while loop untuk input berulang sampai mengetik 'selesai'
while True:
    hari = str(input("\nHari (contoh: Sen, Sel,...) (Ketik 'Selesai' untuk stop): "))

        #lower digunakan untuk mengubah karakter besar menjadi kecil semua (contoh: 'Selesai' ---> 'selesai') agar sama dengan variabel yang diberikan pada kondisi dan terbaca
    if hari.lower() == 'selesai':
        #Break digunakan untuk menghentikan loop secara paksa jika input adalah selesai
        break

    style = str(input("Style (Striker/Grappler) : ")).lower()
    if style == "striker":
        print("Pilihan: Tendangan, Pukulan, Atletik, Weightlifting")
    else:
        print("Pilihan: Takedown, Submission, Atletik, Weightlifting")

    jenis = str(input("Jenis Latihan : ")).lower()
    waktu = int(input("Waktu (menit) : "))

        #Buat data kosong
        #neval berfungsi memberi angka default agar program tidak error jika input di luar pilihan
        #teval berfungsi agar tabel tetap terisi dan rapi jika input tidak dikenali
    neval = 0.0
    teval = "-"
    if jenis in ['tendangan', 'pukulan', 'takedown', 'submission']:
        jumlah = int(input("Jumlah Gerakan: "))
        if waktu > 0:
            neval = jumlah / waktu
        else:
            neval = 0
        teval = f"{neval:.1f}x/min"

    elif jenis == 'atletik':
        jarak = float(input("Jarak (km)    : "))
        if jarak > 0:
            neval = waktu / jarak
        else:
            teval = 0
        teval = f"Pace {neval:.2f}"

    riwayatlatihan.append({'hari': hari, 'style': style, 'jenis': jenis, 'waktu': waktu, 'teval': teval, 'neval': neval})

print("\n")
print("-"*50)

#3. PENGELOLAAN DATA DENGAN STRUKTUR print

print("======Database Tim Olahraga======")
print("-"*50)

    #Menggunakan list comprehension untuk membuat list sementara yang berisi pace lebih dari 0
pacecom = [data['neval'] for data in riwayatlatihan if data['jenis'] ==  'atletik' and data['neval']> 0]
    #Cari nilai pace terkecil dari pacecom menggunakan fungsi min
    #Menggunakan if pacecom untuk mencegah error ketika list kosong
if pacecom:
    paceterbaik = min(pacecom)
else:
    paceterbaik = 0

    #List digunakan untuk data yang bisa diubah (Muteable) dan berurut (Ordered)
    #Buat list kosong
skorfisik = []

while True:
        #Gunakan input untuk memasukkan seberapa banyak data skor fisik
    iskor = input("Masukkan skor fisik anggota (ketik 'Selesai' untuk stop): ")
    if iskor.lower() == 'selesai':
        break

        #Memastikan input berupa angka
    if iskor.isdigit():
        skorfisik.append(int(iskor))
    else:
        print("Harap masukkan angka bulat yang valid.")

    #Mencegah error fungsi max() & sum() jika list kosong
if not skorfisik:
    skorfisik = [0]

    #Mencari skor tertinggi, total, dan rata rata menggunakan fungsi max, sum, dan len
skortinggi = max(skorfisik)
rataskor = sum(skorfisik) / len(skorfisik)

    #Tuple digunakan untuk data yang tidak bisa diubah (Imuteable)
    #Tuple menggunakan simbol kurung biasa ()
atlet1 = ("Ardi", "18", "Striker")
atlet2 = ("Sandi", "18", "Grappler")


    #Mencegah index error jika user memasukkan kurang dari 2 skor
if len(skorfisik) > 0:
    skora = skorfisik[0]
else:
    skora = 0

if len(skorfisik) > 1:
    skorb = skorfisik[1]
else:
    skorb = 0

    #Dictionary digunakan untuk menyimpan profil atlet
    #Dictionary menggunakan simbol kurung kurawal {}
database = {atlet1[0]: {'usia': atlet1[1], 'spesialisasi': atlet1[2], 'skor': skora}, atlet2[0]: {'usia': atlet2[1], 'spesialisasi': atlet2[2], 'skor': skorb}}

    #Set digunakan untuk menyimpan data unik dan menghapus duplikat secara otomatis
    #Set menggunakan simbol kurung kurawal {}, tapi tidak menggunakan kunci
    #Mengambil data latihan unik dari inputan menggunakan set comprehension
    #Jika tidak ada input (langsung ketik selesai), kita beri nilai default agar tidak kosong
latihanardi = {data['jenis'].capitalize() for data in riwayatlatihan}
if not latihanardi:
    latihanardi = {"Belum ada latihan"}

    #Set statis untuk Sandi sebagai pembanding irisan
latihansandi = {"Takedown", "Atletik", "Weightlifting", "Pukulan"}

    #Operasi irisan (intersection) untuk mencari latihan gabungan
latihanirisan = latihanardi & latihansandi

#4. PRINT MENGGUNAKAN F STRING
print("\n\n")
print("="*26)
print("     •OUTPUT PROGRAM•     ")
print("="*26)
print("\n")
print("-"*50)
print("======EVALUASI KONDISI FISIK======")
print("-"*50)

print(f"Detak Jantung Istirahat   : {hristirahat} bpm ---> {sht}")
print(f"Tekanan Darah             : {sistolik}/{diastolik} ---> {std}")
print(f"VO2Max                    : {vo2max}   ---> {svo}")
print(f"IMT                       : {imt:.1f}   ---> {simt}")

print("\n🏋‍♂️ REKOMENDASI LATIHAN")
if sht == "Normal" and std == "Normal":
    print(" ✓ Latihan aerobik intensitas sedang hingga berat (3-5×/minggu")
    print(" ✓ Pertahankan pola makan gizi seimbang untuk menjaga performa")
else:
    print(" ✓ Latihan aerobik intensitas ringan (2-3×/minggu")
    print(" ✓ Kontrol asupan garam/natrium (maksimal 2 mg/hari), pantau tekanan darah")

print("\n")
print("-"*50)
print("====PROSES LATIHAN MMA MINGGUAN====")
print("-"*50)
    #{teks:<angka} berarti teks rata kiri dengan lebar spasi tertentu
print(f"{'Hari':<5} | {'Jenis':<11} | {'Waktu':<5} | {'Evaluasi':<10} | {'Grafik'}")
print("-"*50)

totalwaktu = 0

    #Menggunakan for loop untuk mencetak isi list riwayat latihan
for data in riwayatlatihan:
    totalwaktu += data['waktu']

    #Mencetak blok sesuai nilai eval (pace / gerakan per menit)
    if data['waktu'] == 0:
        grafik = "🛏"
    else:
            #Grafik bertambah berdasarkan neval, jika neval berjumlah 5 maka grafik akan menjadi 5
        grafik = "█" * int(data['neval'])

        if data['jenis'] == 'atletik' and data['neval'] == paceterbaik and paceterbaik > 0:
            grafik += "🏆"
    print(f"{data['hari']:<5} | {data['jenis'].capitalize():<11} | {data['waktu']:>2}min | {data['teval']:<10} | {grafik}")


print("-"*50)
print(f"Total Waktu Latihan Mingguan: {totalwaktu} menit")

print("\n")
print("-"*50)
print("====DATABASE TIM MMA & CABOR PENUNJANG====")
print("-"*50)

print(f"📋 Skor              : {skorfisik}")

    #Menampilkan data dari dictionary menggunakan fungsi .items()
for nama, info in database.items():
    print(f"👤 {nama:<8} ---> Tipe: {info['spesialisasi']:<8} | Skor: {info['skor']} | Usia: {info['usia']}")

print(f"🏅 Latihan Ardi      : {latihanardi}")
print(f"🏅 Latihan Sandi     : {latihansandi}")
print(f"🏅 Latihan Gabungan  : {latihanirisan}")
print(f"📊 Rata-rata Skor    : {rataskor:.2f} | Top Skor: {skortinggi}")
print("-"*50)
```
